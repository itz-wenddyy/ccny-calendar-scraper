```python
# Import packages
from datetime import datetime
from bs4 import BeautifulSoup
import pandas as pd
import requests

# Specify the url
url = "https://www.ccny.cuny.edu/registrar/fall"

# Add a User-Agent header to prevent getting blocked by the server
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}

# Fetch the page
r = requests.get(url, headers=headers)
text = r.text

# Print the first 500 characters to verify
print(text[:500])

# Parse the HTML with BeautifulSoup
soup = BeautifulSoup(text, 'html.parser')

# Print the prettified HTML to verify parsing
print(soup.prettify()[:500])

# Find all tables
tables = soup.find_all('table')

# Check if any tables exist to prevent an index error
if len(tables) > 0:
  table = tables[0]
  # find all rows
  rows = table.find_all('tr')
else:
  rows = []

# loop through all rows, extracting the text from each cell and storing it in a list
data_rows = []
index_dates = []

for row in rows:
  cells = row.find_all('td')

  # Use a standard if statement to ensure the row has enough columns
  if len(cells) >= 3:
    date_str = cells[0].text
    dow_str = cells[1].text
    text_str = cells[2].text

    # Check if the date cell is not empty
    if date_str != "":
      # Add the year 2021 and convert the string into a Python date object
      full_date_str = date_str + ", 2021"
      parsed_date = datetime.strptime(full_date_str, "%B %d, %Y").date()

      # Store the date for the index and the dictionary for the columns
      index_dates.append(parsed_date)
      data_rows.append({'dow': dow_str, 'text': text_str})

# Convert the data into a pandas DataFrame with datetime index and specific columns
df = pd.DataFrame(data_rows, index=index_dates)

# Name the index column
df.index.name = 'Date'

# Print the final cleaned DataFrame
print(df.head())
```
